Transformando dados


```python
import joblib
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
df = pd.read_csv("..\datasets\spotify_limpo.csv", encoding='utf-8', low_memory=True)
```

```python
# 1. Cópia e limpeza inicial
df_transf = df.copy()
df_transf = df_transf.drop(columns=['Unnamed: 0', 'track_id', 'track_name', 'album_name', 'artists', 'track_genre'])
```

```python
# 2. Criar target binário
df_transf['is_popular'] = df_transf['popularity'].apply(lambda x: 1 if x >= 60 else 0)
df_transf = df_transf.drop(columns=['popularity'])
```

```python
colunas_categoricas = ['key', 'mode', 'time_signature']
colunas_numericas = [col for col in df_transf.columns if col not in colunas_categoricas + ['is_popular']]
```

```python
ohe = OneHotEncoder(sparse_output=False, drop='first')
encoded_categoricas = ohe.fit_transform(df_transf[colunas_categoricas])
df_encoded = pd.DataFrame(encoded_categoricas, columns=ohe.get_feature_names_out(colunas_categoricas))
```

```python
# StandardScaler
scaler = StandardScaler()
scaled_numericas = scaler.fit_transform(df_transf[colunas_numericas])
df_scaled = pd.DataFrame(scaled_numericas, columns=colunas_numericas)
```

```python
# 4. Concatenar features (X) e adicionar target (y) como última coluna
df_final = pd.concat([df_scaled, df_encoded, df_transf[['is_popular']]], axis=1)
```

```python
# 5. Salvar dataset processado
df_final.to_csv("../datasets/spotify_processed.csv", index=False, encoding='utf-8')
```

```python
# 6. Salvar os scalers e encoder para uso futuro
joblib.dump(scaler, '../modelos/standard_scaler.joblib')
joblib.dump(ohe, '../modelos/onehot_encoder.joblib')
```
