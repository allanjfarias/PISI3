# Scatter plots

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
from itertools import combinations
import plotly.io as pio

pio.renderers.default = 'browser'
```

```python
df = pd.read_csv('..\datasets\spotify_limpo.csv')

CLUSTER_FEATURES = [
    'danceability',
    'energy',
    'loudness',
    'speechiness',
    'acousticness',
    'valence',
    'tempo'
]
```

```python
feature_pairs = list(combinations(CLUSTER_FEATURES, 2))[:6]  # Você pode aumentar esse número

# Gerar vários plots
figs = []
for x_feature, y_feature in feature_pairs:
    fig = px.scatter(
        df,
        x=x_feature,
        y=y_feature,
        color="cluster" if "cluster" in df.columns else None,
        hover_data=["track_name", "artists", "track_genre"],
        title=f"{x_feature} vs {y_feature}",
        height=500,
        width=600
    )
    fig.show()
```
