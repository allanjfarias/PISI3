# Aplicando Scalers


```python
import joblib
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
df = pd.read_csv("..\\datasets_paralelos\\spotify_limpo.csv", encoding='utf-8', low_memory=True)
```

```python
minmax = MinMaxScaler()
```

```python

df_transf = df.copy()
df_transf = df_transf.drop(columns=['Unnamed: 0', 'track_id', 'track_name', 'album_name', 'artists', 'track_genre'])

print(df_transf.columns)
```

```python
tratar_poucas = ['tempo', 'loudness', 'duration_min']
```

```python

df_normalized = minmax.fit_transform(df[tratar_poucas])
df_normalized = pd.DataFrame(df_normalized, columns=tratar_poucas)   
df_transf[tratar_poucas] = df_normalized[tratar_poucas]
```

```python

df_transf.to_csv(f"..\\datasets_paralelos\\spotify_normalizado.csv", index=False, encoding='utf-8')
```
