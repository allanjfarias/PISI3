# DataCleaning

```python
import pandas as pd 
```

```python
# Carrega o dataset
df = pd.read_csv("..\datasets\spotify_original.csv", encoding='utf-8', low_memory=True)
```

Descobrindo valores Null

```python
# 1. Mostrar quantidade total de valores nulos por coluna
print('Valores nulos por coluna:')
linhas_com_nulos = df[df.isnull().any(axis=1)]
print(linhas_com_nulos)
```

Removendo músicas repetidas do mesmo artista.

```python
# Mantém apenas a primeira linha de cada track_id (uma linha por música)
df_limpo = df.drop_duplicates(subset=['track_name', 'artists'], keep='first')


print(df_limpo.shape)
```

```python
df_limpo.to_csv("..\datasets\spotify_limpo.csv", index=False, encoding='utf-8')
```
